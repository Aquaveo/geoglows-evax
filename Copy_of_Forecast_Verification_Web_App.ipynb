# 1. Introduction

Forecast verification answers a deceptively simple question: **how good is the forecast?** In operational hydrology, this question is especially critical because the answer determines when to issue warnings, how much trust to place in model output, and where to focus model improvement efforts.

This Colab implements a comprehensive verification framework for ECMWF ensemble streamflow forecasts produced by the GEOGLOWS system. The framework evaluates forecasts issued during a specific flood event against observed gauge data, across multiple lead times (0–15 days) and all 52 ensemble members.

## 1.1 Why ensemble verification is different

A deterministic forecast produces a single value. An ensemble forecast produces a **distribution** of possible futures — in this case, 51 perturbed members plus one high-resolution member (ensemble_52, available up to lead day 10). This distinction matters for verification because:

- Some metrics evaluate **each member independently** as a deterministic forecast (MCC, HSS, KGE', peak timing error)
- Other metrics evaluate the **ensemble as a collective distribution** (CRPS)
- The spread between members is itself informative — a wide spread at the right time is desirable

## 1.2 Structure of the verification framework

The metrics implemented here cover four complementary aspects of forecast quality:

| Family | Metrics | Question answered |
|---|---|---|
| Categorical | Contingency matrix, MCC, HSS | Did the forecast correctly classify the event severity? |
| Timing | Peak timing error, Threshold crossing | Did the forecast get the timing right? |
| Accuracy | KGE', r, β, γ | How close was the forecast to the observations in magnitude and shape? |
| Probabilistic | CRPS | Did the ensemble distribution cover the observation? |

No single metric tells the complete story. A model can have good timing but poor magnitude, or good categorical skill but poor probabilistic calibration. Reading these metrics together provides a diagnostic portrait of model behavior.

````python
def gumbel_1(std: float, xbar: float, rp: int or float) -> float:
  """
  Solves the Gumbel Type I probability distribution function (pdf) = exp(-exp(-b)) where b is the covariate. Provide
  the standard deviation and mean of the list of annual maximum flows. Compare scipy.stats.gumbel_r
  Args:
  	std (float): the standard deviation of the series
    xbar (float): the mean of the series
    rp (int or float): the return period in years
  Returns:
  	float, the flow corresponding to the return period specified
  """
  # xbar = statistics.mean(year_max_flow_list)
  # std = statistics.stdev(year_max_flow_list, xbar=xbar)
  return -math.log(-math.log(1 - (1 / rp))) * std * .7797 + xbar - (.45 * std)
````

````python
def reorganize_forecasts_by_daily_lead(forecasts_dict, keep_lead0=True, max_lead=15):
    """
    Reorganize ALL forecast values into daily lead buckets (24h windows),
    using inclusive upper bounds.

    Lead definition:
        Lead 0  -> initialization only (t0)
        Lead 1  -> (0h, 24h]  including 24h
        Lead 2  -> (24h, 48h] including 48h
        Lead 3  -> (48h, 72h]
        ...
    """

    lead_dict = defaultdict(dict)

    lead_days = range(0 if keep_lead0 else 1, max_lead + 1)

    for init_str, data in forecasts_dict.items():

        t0 = pd.to_datetime(init_str, format="%Y%m%d")

        # ---------- 3h ensemble (51) ----------
        t3 = pd.DatetimeIndex(data["time_3h"])
        a3 = data["ens_3h"]  # (n, 51)

        # lead day for each timestamp
        lead3 = np.ceil(((t3 - t0).to_numpy() / np.timedelta64(1, "h")) / 24.0).astype(int)

        if keep_lead0:
            lead3 = lead3.copy()
            lead3[(t3 == t0)] = 0

        # ---------- 1h high-res (1) ----------
        t1 = pd.DatetimeIndex(data["time_1h"])
        a1 = data["ens_1h"]  # (m, 1)

        lead1 = np.ceil(((t1 - t0).to_numpy() / np.timedelta64(1, "h")) / 24.0).astype(int)

        if keep_lead0:
            lead1 = lead1.copy()
            lead1[(t1 == t0)] = 0

        for d in lead_days:
            # subset 3h
            mask3 = (lead3 == d)
            sub_t3 = t3[mask3]
            sub_a3 = a3[mask3, :] if mask3.any() else None

            # subset 1h
            mask1 = (lead1 == d)
            sub_t1 = t1[mask1]
            sub_a1 = a1[mask1, :] if mask1.any() else None

            # Only save if there is something (3h o 1h)
            #if (sub_a3 is not None and len(sub_t3) > 0) or (sub_a1 is not None and len(sub_t1) > 0):
            if (sub_a3 is not None) or (sub_a1 is not None):
                lead_dict[d][init_str] = {"time_3h": sub_t3, "ens_3h": sub_a3, "time_1h": sub_t1, "ens_1h": sub_a1}

    return dict(lead_dict)
````

````python
def to_hourly(time_index, values_array):
    """
    Convierte una serie (o múltiples columnas) a resolución horaria,
    preservando todos los valores y usando interpolación lineal.

    Parameters
    ----------
    time_index : DatetimeIndex
    values_array : numpy array (n, m)

    Returns
    -------
    new_time_index : DatetimeIndex horario
    new_values : numpy array horario (k, m)
    """

    # Crear DataFrame temporal
    df = pd.DataFrame(values_array, index=pd.DatetimeIndex(time_index))

    # Crear índice horario completo
    full_index = pd.date_range(
        start=df.index.min(),
        end=df.index.max(),
        freq="h"
    )

    # Reindexar
    df = df.reindex(full_index)

    # Interpolar lineal
    df = df.interpolate(method="linear")

    return df.index, df.to_numpy(dtype=np.float32)
````

````python
def get_hourly_resolution(event_data_df):
    """
    Ensures that event_data_df has hourly temporal resolution.

    Rules:
    - If already hourly → return unchanged
    - If sub-hourly → aggregate to hourly using mean
    - If coarser than hourly → resample to hourly and interpolate
    """

    df = event_data_df.copy()
    df = df.sort_index()

    # ---------------------------------------
    # 1️⃣ Estimate temporal resolution
    # ---------------------------------------
    inferred_freq = pd.infer_freq(df.index)

    if inferred_freq is None:
        print("Frequency could not be inferred automatically.")
        time_diffs = df.index.to_series().diff().dropna()
        median_diff = time_diffs.median()
    else:
        print(f"Detected frequency: {inferred_freq}")
        median_diff = pd.to_timedelta(pd.tseries.frequencies.to_offset(inferred_freq))

    print(f"Estimated median time step: {median_diff}")

    one_hour = pd.Timedelta(hours=1)

    # ---------------------------------------
    # 2️⃣ Already hourly
    # ---------------------------------------
    if abs(median_diff - one_hour) < pd.Timedelta(minutes=1):
        print("Data are already hourly. No resampling applied.")
        return df

    # ---------------------------------------
    # 3️⃣ Sub-hourly (e.g., 15min, 30min)
    # ---------------------------------------
    elif median_diff < one_hour:
        print("Data are sub-hourly. Aggregating to hourly using mean.")
        df_hourly = df.resample("H").mean()
        return df_hourly

    # ---------------------------------------
    # 4️⃣ Coarser than hourly (e.g., 3h, daily)
    # ---------------------------------------
    else:
        print("Data are coarser than hourly. Resampling to hourly using linear interpolation.")

        # Create full hourly index
        full_index = pd.date_range(start=df.index.min(), end=df.index.max(), freq="H")

        df_hourly = df.reindex(full_index)

        # Interpolate linearly
        df_hourly = df_hourly.interpolate(method="linear")

        # Fill possible edge NaNs
        df_hourly = df_hourly.bfill().ffill()

        return df_hourly
````

````python
def build_lead_dataframe(lead_forecasts_dict, lead_day):
    """
    Construye un DataFrame único para un lead_day dado,
    concatenando todas las fechas de inicialización.

    Elimina ensemble_52 si todos sus valores son NaN.
    """

    lead_block = lead_forecasts_dict[lead_day]

    dfs = []

    for init_date, data in lead_block.items():

        # ----- ENSAMBLE 51 miembros -----
        df_ens = pd.DataFrame(
            data["ens_3h"],
            index=pd.DatetimeIndex(data["time_3h"]),
            columns=[f"ensemble_{str(i+1).zfill(2)}" for i in range(51)]
        )

        # ----- HIGH RES (miembro 52) -----
        df_hr = pd.DataFrame(
            data["ens_1h"],
            index=pd.DatetimeIndex(data["time_1h"]),
            columns=["ensemble_52"]
        )

        # ----- UNIR POR DATETIME (outer join) -----
        df = df_ens.join(df_hr, how="outer")

        dfs.append(df)

    # ----- CONCATENAR TODAS LAS INICIALIZACIONES -----
    final_df = pd.concat(dfs)

    # ----- ORDENAR POR FECHA -----
    final_df = final_df.sort_index()

    # ----- ELIMINAR ensemble_52 SI ES TODO NaN -----
    if "ensemble_52" in final_df.columns:
        if final_df["ensemble_52"].isna().all():
            final_df = final_df.drop(columns=["ensemble_52"])

    return final_df
````

````python
def compute_lead_forecast_stats(lead_forecasts_dict, max_lead=15):

    lead_stats_dict = {}

    for lead_day in range(0, max_lead + 1):

        if lead_day not in lead_forecasts_dict:
            continue

        lead_stats_dict[lead_day] = {}

        for init_str, data in lead_forecasts_dict[lead_day].items():

            ens = data["ens_3h"]   # (n, 51)
            time_index = pd.DatetimeIndex(data["time_3h"])

            # ---------------------------
            # LEAD 0 → only flow_avg
            # ---------------------------
            if lead_day == 0:

                # All ensemble members are identical
                flow_avg = ens[:, 0]  # take first column

                lead_stats_dict[lead_day][init_str] = {
                    "time": time_index,
                    "stats": flow_avg.astype(np.float32),
                    "columns": ["flow_avg"]
                }

            # ---------------------------
            # LEAD 1–10
            # ---------------------------
            elif 1 <= lead_day <= 10:

                flow_min  = np.nanmin(ens, axis=1)
                flow_25p  = np.nanpercentile(ens, 25, axis=1)
                flow_avg  = np.nanmean(ens, axis=1)
                flow_med  = np.nanpercentile(ens, 50, axis=1)
                flow_75p  = np.nanpercentile(ens, 75, axis=1)
                flow_max  = np.nanmax(ens, axis=1)

                stats_array = np.column_stack([
                    flow_min,
                    flow_25p,
                    flow_avg,
                    flow_med,
                    flow_75p,
                    flow_max
                ]).astype(np.float32)

                # high resolution
                high_res = data["ens_1h"]

                lead_stats_dict[lead_day][init_str] = {
                    "time": time_index,
                    "stats": stats_array,
                    "columns": ["flow_min", "flow_25p", "flow_avg", "flow_med", "flow_75p", "flow_max"],
                    "high_res": high_res.astype(np.float32)
                }

            # ---------------------------
            # LEAD 11–15
            # ---------------------------
            else:

                flow_min  = np.nanmin(ens, axis=1)
                flow_25p  = np.nanpercentile(ens, 25, axis=1)
                flow_avg  = np.nanmean(ens, axis=1)
                flow_med  = np.nanpercentile(ens, 50, axis=1)
                flow_75p  = np.nanpercentile(ens, 75, axis=1)
                flow_max  = np.nanmax(ens, axis=1)

                stats_array = np.column_stack([
                    flow_min,
                    flow_25p,
                    flow_avg,
                    flow_med,
                    flow_75p,
                    flow_max
                ]).astype(np.float32)

                lead_stats_dict[lead_day][init_str] = {
                    "time": time_index,
                    "stats": stats_array,
                    "columns": ["flow_min", "flow_25p", "flow_avg", "flow_med", "flow_75p", "flow_max"]
                }

    return lead_stats_dict
````

````python
def build_lead_stats_dataframe(lead_forecast_stats_dict, lead_day):
    """
    Builds a single DataFrame for a given lead_day
    from lead_forecast_stats_dict.

    Handles:
    - Lead 0 (only flow_avg)
    - Lead 1–10 (stats + high_res)
    - Lead 11–15 (stats only)
    """

    if lead_day not in lead_forecast_stats_dict:
        raise ValueError(f"Lead day {lead_day} not found.")

    lead_block = lead_forecast_stats_dict[lead_day]

    dfs = []

    for init_str, data in lead_block.items():

        time_index = pd.DatetimeIndex(data["time"])

        # -------- Build stats dataframe --------
        if lead_day == 0:
            # stats is 1D array
            df_stats = pd.DataFrame(
                data["stats"],
                index=time_index,
                columns=["flow_avg"]
            )
        else:
            df_stats = pd.DataFrame(
                data["stats"],
                index=time_index,
                columns=data["columns"]
            )

        # -------- Add high_res if present --------
        if "high_res" in data:
            df_hr = pd.DataFrame(
                data["high_res"],
                index=time_index,
                columns=["high_res"]
            )
            df = df_stats.join(df_hr, how="outer")
        else:
            df = df_stats

        dfs.append(df)

    # -------- Concatenate all initializations --------
    final_df = pd.concat(dfs)

    # -------- Sort by datetime --------
    final_df = final_df.sort_index()

    # -------- Remove high_res if fully NaN --------
    if "high_res" in final_df.columns:
        if final_df["high_res"].isna().all():
            final_df = final_df.drop(columns=["high_res"])

    return final_df
````

````python
def determine_event_return_period(hourly_df, observed_rp, comid):
    """
    Determines the maximum observed return period exceeded during the event.
    """

    # Extract thresholds
    thresholds = {
        2: observed_rp.loc[2, comid],
        5: observed_rp.loc[5, comid],
        10: observed_rp.loc[10, comid],
        25: observed_rp.loc[25, comid],
        50: observed_rp.loc[50, comid],
        100: observed_rp.loc[100, comid],
    }

    max_observed_flow = hourly_df.iloc[:, 0].max()

    event_return_period = 0

    for rp in sorted(thresholds.keys()):
        if max_observed_flow >= thresholds[rp]:
            event_return_period = rp

    print(f"The observed event exceeds the {event_return_period}-year return period.")

    return event_return_period
````

````python
def classify_series(series, thresholds, event_return_period):
    """
    Classifies series into interval-based return period categories up to the observed event magnitude.
    """

    categories = pd.Series(index=series.index, dtype=int)

    rp_levels = [2,5,10,25,50,100]
    rp_levels = [rp for rp in rp_levels if rp <= event_return_period]

    # Below 2-year
    categories[series < thresholds[2]] = 0

    for i, rp in enumerate(rp_levels):

        if rp == 100:
            categories[series >= thresholds[100]] = 100
        else:
            next_rp = rp_levels[i+1]
            categories[(series >= thresholds[rp]) & (series < thresholds[next_rp])] = rp

    return categories
````

````python
# ===========================================================================
# Function: deterministic contingency matrix
# Replaces build_probabilistic_contingency_matrix
# ===========================================================================

def build_deterministic_contingency_matrix(
    forecast_series,          # pd.Series  — selected forecast series
    observed_df,              # pd.DataFrame — observed streamflow (1 column)
    observed_rperiods_df,     # pd.DataFrame — observed thresholds (index=RP, col=comid)
    simulated_rperiods_df,    # pd.DataFrame — simulated thresholds (index=RP, col=comid)
    comid,
    event_return_period
):
    """
    Builds a deterministic contingency matrix for a single forecast series
    (one ensemble member or a derived statistic such as the median).

    Unlike build_probabilistic_contingency_matrix, this function does not
    average across members. It compares one series directly against
    observations on a timestep-by-timestep basis.

    Parameters
    ----------
    forecast_series : pd.Series
        Forecast series with a DatetimeIndex (e.g., lead_df["ensemble_01"]
        or an externally computed statistic).
    observed_df : pd.DataFrame
        Observed streamflow during the event at hourly resolution.
    observed_rperiods_df : pd.DataFrame
        Return period thresholds derived from historical observed data.
        Index = return period (2, 5, 10, 25, 50, 100), column = comid.
    simulated_rperiods_df : pd.DataFrame
        Return period thresholds derived from the retrospective simulation.
        Index = return period (2, 5, 10, 25, 50, 100), column = comid.
    comid : int
        Reach identifier (river_id).
    event_return_period : int
        Maximum return period exceeded during the event. Determines the
        number of categories in the matrix.

    Returns
    -------
    matrix : pd.DataFrame
        Normalized contingency matrix (each cell = fraction of timesteps).
        Index and columns are the valid return period categories [0, 2, 5, ...].
    summary : dict
        {"hits", "underestimation", "overestimation", "n_timesteps"}
    """

    # ---------------------------------
    # 1. Align temporal window
    # ---------------------------------
    start = max(forecast_series.index.min(), observed_df.index.min())
    end   = min(forecast_series.index.max(), observed_df.index.max())

    fcst = forecast_series.loc[start:end]
    obs  = observed_df.loc[start:end]

    # ---------------------------------
    # 2. Threshold dictionaries
    # ---------------------------------
    obs_thresholds = {rp: observed_rperiods_df.loc[rp, comid]
                      for rp in [2, 5, 10, 25, 50, 100]}
    sim_thresholds = {rp: simulated_rperiods_df.loc[rp, comid]
                      for rp in [2, 5, 10, 25, 50, 100]}

    # ---------------------------------
    # 3. Classify observations and forecast
    # ---------------------------------
    obs_cat = classify_series(obs.iloc[:, 0], obs_thresholds, event_return_period)

    # Keep only timesteps where both series have valid data
    aligned = pd.concat([obs_cat, fcst], axis=1).dropna()
    if aligned.empty:
        raise ValueError("No overlapping timesteps between forecast and observations.")

    obs_aligned  = aligned.iloc[:, 0]
    fcst_aligned = classify_series(aligned.iloc[:, 1], sim_thresholds, event_return_period)

    # ---------------------------------
    # 4. Valid categories (dynamic based on event magnitude)
    # ---------------------------------
    valid_categories = [0] + [rp for rp in [2, 5, 10, 25, 50, 100]
                               if rp <= event_return_period]

    # ---------------------------------
    # 5. Build and normalize the matrix
    # ---------------------------------
    n_timesteps = len(aligned)
    matrix = pd.DataFrame(0, index=valid_categories, columns=valid_categories, dtype=float)

    for obs_c in valid_categories:
        for fcst_c in valid_categories:
            count = np.sum((obs_aligned == obs_c) & (fcst_aligned == fcst_c))
            matrix.loc[obs_c, fcst_c] = count

    #matrix = matrix / n_timesteps   # each cell = fraction of timesteps

    # ---------------------------------
    # 6. Error decomposition
    # ---------------------------------
    hits            = np.trace(matrix.values)
    underestimation = matrix.where(
        np.tril(np.ones(matrix.shape), k=-1).astype(bool)).sum().sum()
    overestimation  = matrix.where(
        np.triu(np.ones(matrix.shape), k=1).astype(bool)).sum().sum()

    #print(f"  Timesteps analyzed : {n_timesteps}")
    #print(f"  Hits               : {hits:.4f}")
    #print(f"  Underestimation    : {underestimation:.4f}")
    #print(f"  Overestimation     : {overestimation:.4f}")
    #print(f"  Total              : {hits + underestimation + overestimation:.4f}")
    #print(f"\n")

    summary = {
        "hits"           : hits,
        "underestimation": underestimation,
        "overestimation" : overestimation,
        "n_timesteps"    : n_timesteps,
    }

    return matrix, summary
````

````python
def get_forecast_options(lead_df):
    """
    Builds dropdown options based on the columns available in lead_df.
    Returns a list of (label, key) tuples for use with widgets.Dropdown.
    """
    options = []

    ensemble_cols = [c for c in lead_df.columns if c.startswith("ensemble_")]

    # Individual ensemble members (01–51)
    member_cols = [c for c in ensemble_cols if c != "ensemble_52"]
    for col in sorted(member_cols):
        num = col.split("_")[1]
        options.append((f"Ensemble {num}", col))

    # High-resolution member (52) if available
    if "ensemble_52" in lead_df.columns:
        options.append(("High-Res (ensemble 52)", "ensemble_52"))

    # Ensemble statistics computed across the 51 members
    if len(member_cols) > 0:
        options += [
            ("── Statistics ────────────", None),   # visual separator (not selectable)
            ("Median (p50)", "stat_median"),
            ("Mean",         "stat_mean"),
            ("Percentile 25","stat_p25"),
            ("Percentile 75","stat_p75"),
            ("Minimum",      "stat_min"),
            ("Maximum",      "stat_max"),
        ]

    return options


def resolve_forecast_series(lead_df, key):
    """
    Given a dropdown key, returns the corresponding pd.Series.
    Statistics are computed on-the-fly across the 51 ensemble members.
    """
    member_cols = [c for c in lead_df.columns
                   if c.startswith("ensemble_") and c != "ensemble_52"]

    if key in lead_df.columns:
        return lead_df[key]

    ens = lead_df[member_cols]

    stat_map = {
        "stat_median": ens.median(axis=1),
        "stat_mean"  : ens.mean(axis=1),
        "stat_p25"   : ens.quantile(0.25, axis=1),
        "stat_p75"   : ens.quantile(0.75, axis=1),
        "stat_min"   : ens.min(axis=1),
        "stat_max"   : ens.max(axis=1),
    }

    if key in stat_map:
        return stat_map[key]

    raise ValueError(f"Key '{key}' not recognized. Valid options: "
                     f"{list(lead_df.columns) + list(stat_map.keys())}")
````

````python
def format_contingency_matrix(matrix, event_return_period):

    rp_labels = ["<2yr"]

    if event_return_period >= 2: rp_labels.append("2–5yr")
    if event_return_period >= 5: rp_labels.append("5–10yr")
    if event_return_period >= 10: rp_labels.append("10–25yr")
    if event_return_period >= 25: rp_labels.append("25–50yr")
    if event_return_period >= 50: rp_labels.append("50–100yr")
    if event_return_period >= 100: rp_labels.append("≥100yr")

    matrix.index = rp_labels
    matrix.columns = rp_labels

    matrix.index.name = "Observed category (↓)"
    matrix.columns.name = "Forecast category (→)"

    return matrix
````

````python
# ===========================================================================
# Function: compute_deterministic_limit
# Changes vs original:
#   - Accepts forecast_series_key to evaluate a specific member or statistic
#   - Uses build_deterministic_contingency_matrix instead of probabilistic
#   - Auto-detects the actual forecast horizon for the selected series
#     (e.g. ensemble_52 is only available up to lead 10)
#   - Removes the break — scans all valid leads and keeps the LAST one where
#     hits > misses + false alarms (faithful to Hewson 2007)
# ===========================================================================

def compute_deterministic_limit(lead_forecasts_dict, event_data_df, observed_rperiods_df, simulated_rperiods_df, comid, forecast_series_key, max_lead=15):
    """
    Computes the deterministic limit following Hewson (2007):
    the last lead day at which hits > misses + false alarms (CSI >= 0.5),
    evaluated for a single forecast series (one ensemble member or statistic).

    The function auto-detects the actual horizon of the selected series.
    For example, ensemble_52 (high-resolution) is only available up to lead 10,
    so the plot and evaluation are limited to that horizon rather than showing
    zeros for the remaining leads.

    Parameters
    ----------
    lead_forecasts_dict : dict
        Output of reorganize_forecasts_by_daily_lead.
    event_data_df : pd.DataFrame
        Observed streamflow during the event.
    observed_rperiods_df : pd.DataFrame
        Observed return period thresholds.
    simulated_rperiods_df : pd.DataFrame
        Simulated return period thresholds.
    comid : int
        Reach identifier.
    forecast_series_key : str
        Key identifying the forecast series to evaluate
        (e.g. "ensemble_01", "ensemble_52", "stat_median").
    max_lead : int
        Maximum lead day to consider (default 15). The actual horizon used
        will be capped at the last lead where the series has valid data.

    Returns
    -------
    deterministic_limit : int
        Last lead day where hits > misses + false alarms.
        Returns -1 if the condition is never met.
    days : list[int]
    hits_rates : list[float]
    misses_rates : list[float]
    """

    event_return_period = determine_event_return_period(
        event_data_df, observed_rperiods_df, comid)

    # -------------------------------------------------------------------------
    # 1. Auto-detect the actual horizon for the selected series
    #    Walk through leads and stop at the first one where the series
    #    is no longer available (KeyError or ValueError from resolve).
    # -------------------------------------------------------------------------
    actual_max_lead = 0

    for lead in range(0, max_lead + 1):
        if lead not in lead_forecasts_dict:
            break
        try:
            lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
            resolve_forecast_series(lead_df, forecast_series_key)
            actual_max_lead = lead
        except (ValueError, KeyError):
            break

    print(f"  Forecast horizon for '{forecast_series_key}': {actual_max_lead} days")

    # -------------------------------------------------------------------------
    # 2. Evaluate hits and misses up to the actual horizon
    # -------------------------------------------------------------------------
    hits_rates   = []
    misses_rates = []
    days         = list(range(0, actual_max_lead + 1))

    for lead in days:

        try:
            lead_df         = build_lead_dataframe(lead_forecasts_dict, lead)
            forecast_series = resolve_forecast_series(lead_df, forecast_series_key)

            matrix, summary = build_deterministic_contingency_matrix(
                forecast_series       = forecast_series,
                observed_df           = event_data_df,
                observed_rperiods_df  = observed_rperiods_df,
                simulated_rperiods_df = simulated_rperiods_df,
                comid                 = comid,
                event_return_period   = event_return_period,
            )

            hits   = summary["hits"]
            misses = summary["underestimation"] + summary["overestimation"]

        except (ValueError, KeyError):
            hits   = 0
            misses = 0

        hits_rates.append(hits)
        misses_rates.append(misses)

    # -------------------------------------------------------------------------
    # 3. Find the LAST lead where hits > misses (no break — full scan)
    #    Faithful to Hewson (2007): the deterministic limit is a
    #    population-level measure, not the first crossing of a noisy curve.
    # -------------------------------------------------------------------------
    deterministic_limit = -1

    for i in range(len(days)):
        if hits_rates[i] > misses_rates[i]:
            deterministic_limit = days[i]   # keep updating — no break

    print(f"  Deterministic limit ({forecast_series_key}): {deterministic_limit} days\n")

    return deterministic_limit, days, hits_rates, misses_rates
````

````python
def compute_mcc(matrix):
    """
    Multi-category Matthews Correlation Coefficient.
    Gorodkin (2004) / Jurman, Riccadonna & Furlanello (2012).

    Accepts a DataFrame or array of raw integer counts (NOT normalized).
    Returns a float in [-1, 1], or np.nan if the matrix is all zeros.
    """
    C   = np.array(matrix, dtype=float)
    s   = C.sum()
    if s == 0:
        return np.nan
    c   = np.trace(C)
    t_k = C.sum(axis=1)
    p_k = C.sum(axis=0)
    numerator = c * s - np.dot(t_k, p_k)
    denom = np.sqrt(
        (s**2 - np.dot(p_k, p_k)) *
        (s**2 - np.dot(t_k, t_k))
    )
    if denom == 0:
        return 0.0
    return float(numerator / denom)
````

````python
def compute_mcc_all_members(lead_forecasts_dict, event_data_df, observed_rperiods_df, simulated_rperiods_df, comid, max_lead=15):
    """
    Computes the multi-category MCC for every ensemble member and
    ensemble statistic at every lead day.

    ensemble_52 is only available up to lead 10 — np.nan is recorded
    for leads 11–15 for that series.

    Returns
    -------
    mcc_results : dict  {lead: {series_key: mcc_value}}
    all_keys    : list  all series keys evaluated
    """
    event_return_period = determine_event_return_period(
        event_data_df, observed_rperiods_df, comid)

    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    mcc_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing MCC per lead"):

        if lead not in lead_forecasts_dict:
            mcc_results[lead] = {k: np.nan for k in all_keys}
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        mcc_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                matrix, _ = build_deterministic_contingency_matrix(
                    forecast_series       = forecast_series,
                    observed_df           = event_data_df,
                    observed_rperiods_df  = observed_rperiods_df,
                    simulated_rperiods_df = simulated_rperiods_df,
                    comid                 = comid,
                    event_return_period   = event_return_period,
                )
                mcc_results[lead][key] = compute_mcc(matrix)
            except (ValueError, KeyError):
                mcc_results[lead][key] = np.nan

    return mcc_results, all_keys
````

````python
def get_member_mcc(lead, include_hires):
    keys  = member_keys + ([hires_key] if include_hires else [])
    vals  = [mcc_results[lead][k] for k in keys
             if not np.isnan(mcc_results[lead].get(k, np.nan))]
    names = [k for k in keys
             if not np.isnan(mcc_results[lead].get(k, np.nan))]
    return np.array(vals), names

def mcc_distribution_stats(leads_list, include_hires_fn):
    meds, q25_l, q75_l, mins_l, maxs_l = [], [], [], [], []
    for l in leads_list:
        vals, _ = get_member_mcc(l, include_hires_fn(l))
        if len(vals) == 0:
            for lst in [meds, q25_l, q75_l, mins_l, maxs_l]:
                lst.append(np.nan)
        else:
            meds.append(float(np.median(vals)))
            q25_l.append(float(np.percentile(vals, 25)))
            q75_l.append(float(np.percentile(vals, 75)))
            mins_l.append(float(np.min(vals)))
            maxs_l.append(float(np.max(vals)))
    return meds, q25_l, q75_l, mins_l, maxs_l
````

````python
def compute_hss(matrix):
    """
    Computes the multi-category Heidke Skill Score (HSS).

    HSS measures the fraction of correct forecasts after removing
    those that would be correct by chance, relative to the maximum
    possible improvement over chance.

        HSS = (N·c  −  Σ t_k · p_k)
              ─────────────────────────────
              (N²  −  Σ t_k · p_k)

    Note: the numerator is identical to the MCC numerator. Only the
    denominator differs — HSS uses N² − Σ t_k·p_k while MCC uses
    √((N² − Σ p_k²)(N² − Σ t_k²)).

    Parameters
    ----------
    matrix : pd.DataFrame or np.ndarray
        K × K contingency matrix with raw integer counts.

    Returns
    -------
    float
        HSS in (-∞, 1]. Returns np.nan if the matrix is all zeros,
        0.0 if denominator is zero (all forecasts in one category).

    References
    ----------
    Wilks, D.S. (2011). Statistical Methods in the Atmospheric Sciences,
        3rd ed. Academic Press. Chapter 8.
    WMO/WWRP Joint Working Group on Forecast Verification Research.
        https://www.cawcr.gov.au/projects/verification/
    """
    C   = np.array(matrix, dtype=float)
    s   = C.sum()
    if s == 0:
        return np.nan
    c      = np.trace(C)
    t_k    = C.sum(axis=1)
    p_k    = C.sum(axis=0)
    dot_tp = np.dot(t_k, p_k)
    numerator   = c * s - dot_tp
    denominator = s**2 - dot_tp
    if denominator == 0:
        return 0.0
    return float(numerator / denominator)
````

````python
def compute_hss_all_members(lead_forecasts_dict, event_data_df, observed_rperiods_df, simulated_rperiods_df, comid, max_lead=15):
    """
    Computes the multi-category HSS for every ensemble member and
    ensemble statistic at every lead day.

    ensemble_52 (high-resolution) is only available up to lead 10.
    For leads 11–15 it is recorded as np.nan.

    Returns
    -------
    hss_results : dict  {lead: {series_key: hss_value}}
    all_keys    : list  all series keys evaluated
    """
    event_return_period = determine_event_return_period(
        event_data_df, observed_rperiods_df, comid)

    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    hss_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing HSS per lead"):

        if lead not in lead_forecasts_dict:
            hss_results[lead] = {k: np.nan for k in all_keys}
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        hss_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                matrix, _ = build_deterministic_contingency_matrix(
                    forecast_series       = forecast_series,
                    observed_df           = event_data_df,
                    observed_rperiods_df  = observed_rperiods_df,
                    simulated_rperiods_df = simulated_rperiods_df,
                    comid                 = comid,
                    event_return_period   = event_return_period,
                )
                hss_results[lead][key] = compute_hss(matrix)
            except (ValueError, KeyError):
                hss_results[lead][key] = np.nan

    return hss_results, all_keys
````

````python
def display_hss_breakdown(lead, forecast_key):
    """
    Rebuilds the contingency matrix for the selected lead and series,
    then shows every intermediate value of the multi-category HSS formula:

        HSS = (N·c  −  Σ t_k·p_k)
              ────────────────────────
              (N²  −  Σ t_k·p_k)

    where:
        C    = contingency matrix (raw counts)
        N    = C.sum()           — total timesteps
        c    = trace(C)          — total hits (diagonal sum)
        t_k  = C.sum(axis=1)    — observed count per category (row sums)
        p_k  = C.sum(axis=0)    — forecast count per category (col sums)

    Note: PC = c/N (proportion correct), PC_ref = Σt_k·p_k / N²
    (proportion correct expected by chance), so:
        HSS = (PC − PC_ref) / (1 − PC_ref)
    """

    # ── 1. Rebuild contingency matrix ────────────────────────────────────────
    event_return_period = determine_event_return_period(
        event_data_df, observed_rperiods_df, comid)

    lead_df         = build_lead_dataframe(lead_forecasts_dict, lead)
    forecast_series = resolve_forecast_series(lead_df, forecast_key)

    matrix, summary = build_deterministic_contingency_matrix(
        forecast_series       = forecast_series,
        observed_df           = event_data_df,
        observed_rperiods_df  = observed_rperiods_df,
        simulated_rperiods_df = simulated_rperiods_df,
        comid                 = comid,
        event_return_period   = event_return_period,
    )

    labeled = format_contingency_matrix(matrix.copy(), event_return_period)
    series_label = next(
        (lbl for lbl, key in selectable_options if key == forecast_key),
        forecast_key
    )

    print("=" * 60)
    print(f"HSS Inspector")
    print(f"  Lead day         : {lead}")
    print(f"  Forecast series  : {series_label}  [{forecast_key}]")
    print(f"  Event RP         : {event_return_period}-yr")
    print("=" * 60)

    # ── 2. Show contingency matrix ────────────────────────────────────────────
    print("\nContingency matrix (raw timestep counts):")
    display(HTML(labeled.to_html()))

    # ── 3. Formula components ─────────────────────────────────────────────────
    C      = np.array(matrix, dtype=float)
    s      = C.sum()                    # N
    c      = np.trace(C)               # hits
    t_k    = C.sum(axis=1)             # row sums
    p_k    = C.sum(axis=0)             # col sums
    dot_tp = np.dot(t_k, p_k)         # Σ t_k · p_k

    pc      = c / s                    # proportion correct
    pc_ref  = dot_tp / s**2            # expected by chance
    numerator   = c * s - dot_tp
    denominator = s**2 - dot_tp
    hss         = numerator / denominator if denominator != 0 else 0.0

    cats = labeled.index.tolist()

    print("\nFormula components:")
    print(f"  N  = total timesteps                    = {int(s)}")
    print(f"  c  = trace(C)  [hits]                   = {int(c)}")
    print(f"  PC = c / N  [proportion correct]        = {pc:.4f}")
    print()
    print(f"  t_k  (observed counts per category):")
    for cat, val in zip(cats, t_k):
        print(f"       {cat:<12}: {int(val):>4}")
    print()
    print(f"  p_k  (forecast counts per category):")
    for cat, val in zip(cats, p_k):
        print(f"       {cat:<12}: {int(val):>4}")
    print()
    print(f"  Σ t_k · p_k                             = {dot_tp:.1f}")
    print(f"  PC_ref = Σt_k·p_k / N²  [chance]       = {pc_ref:.4f}")
    print()
    print(f"  Numerator   = N·c − Σ t_k·p_k          = {s:.0f}·{c:.0f} − {dot_tp:.1f} = {numerator:.1f}")
    print(f"  Denominator = N² − Σ t_k·p_k           = {s**2:.1f} − {dot_tp:.1f} = {denominator:.1f}")
    print()
    print(f"  Alternative form:")
    print(f"    HSS = (PC − PC_ref) / (1 − PC_ref)")
    print(f"        = ({pc:.4f} − {pc_ref:.4f}) / (1 − {pc_ref:.4f})")
    print(f"        = {pc - pc_ref:.4f} / {1 - pc_ref:.4f}")
    print()
    print(f"  ┌─────────────────────────────────────┐")
    print(f"  │  HSS = {numerator:.1f} / {denominator:.1f} = {hss:.6f}  │")
    print(f"  └─────────────────────────────────────┘")

    # Cross-check with MCC
    from math import sqrt
    sum_p2 = np.dot(p_k, p_k)
    sum_t2 = np.dot(t_k, t_k)
    mcc_denom = sqrt((s**2 - sum_p2) * (s**2 - sum_t2)) if (s**2-sum_p2)*(s**2-sum_t2) > 0 else 0
    mcc = numerator / mcc_denom if mcc_denom != 0 else 0.0

    #print(f"\n  Note: numerator is identical to MCC numerator = {numerator:.1f}")
    #print(f"  MCC denominator = {mcc_denom:.4f}  →  MCC = {mcc:.6f}")
    print(f"\n  HSS numerator = {numerator:.1f}")
    print(f"  HSS denominator = {denominator:.4f}  →  HSS = {hss:.6f}")

    stored = hss_results.get(lead, {}).get(forecast_key, np.nan)
    #if not np.isnan(stored):
    #    print(f"\n  Stored in hss_results : {stored:.6f}")
````

````python
# ===========================================================================
# Peak Timing Error Inspector
# Shows the Δt_peak calculation for the currently selected
# lead day (selected_lead) and forecast series (selected_forecast_key).
#
# Definition:
#   Δt_peak = t_peak_forecast − t_peak_observed   [hours]
#   Negative → forecast peak arrives EARLY
#   Positive → forecast peak arrives LATE
#   Zero     → perfect timing
#
# Window: only the overlapping period between forecast and observations.
# Peak:   the timestep of maximum streamflow within that window.
# ===========================================================================

def compute_peak_timing_error(forecast_series, observed_df):
    """
    Computes the peak timing error (hours) between a forecast series
    and the observed streamflow, within their overlapping time window.

    Parameters
    ----------
    forecast_series : pd.Series
        Forecast streamflow with DatetimeIndex.
    observed_df : pd.DataFrame
        Observed streamflow with DatetimeIndex (1 column).

    Returns
    -------
    delta_t_hours : float
        t_peak_forecast − t_peak_observed in hours.
        Negative = early, positive = late.
    t_peak_obs : pd.Timestamp
    t_peak_fcst : pd.Timestamp
    q_peak_obs : float
    q_peak_fcst : float
    """
    # Overlapping window
    start = max(forecast_series.index.min(), observed_df.index.min())
    end   = min(forecast_series.index.max(), observed_df.index.max())

    if start >= end:
        raise ValueError("No overlapping window between forecast and observations.")

    fcst_window = forecast_series.loc[start:end].dropna()
    obs_window  = observed_df.iloc[:, 0].loc[start:end].dropna()

    if fcst_window.empty or obs_window.empty:
        raise ValueError("Empty series after restricting to overlapping window.")

    t_peak_obs  = obs_window.idxmax()
    t_peak_fcst = fcst_window.idxmax()
    q_peak_obs  = obs_window.max()
    q_peak_fcst = fcst_window.max()

    delta_t = (t_peak_fcst - t_peak_obs).total_seconds() / 3600.0

    return delta_t, t_peak_obs, t_peak_fcst, q_peak_obs, q_peak_fcst


def display_peak_timing_breakdown(lead, forecast_key):
    """
    Shows the peak timing error calculation for the selected
    lead day and forecast series.
    """
    lead_df         = build_lead_dataframe(lead_forecasts_dict, lead)
    forecast_series = resolve_forecast_series(lead_df, forecast_key)

    series_label = next(
        (lbl for lbl, key in selectable_options if key == forecast_key),
        forecast_key
    )

    try:
        delta_t, t_obs, t_fcst, q_obs, q_fcst = compute_peak_timing_error(
            forecast_series, event_data_df)
    except ValueError as e:
        print(f"Cannot compute peak timing error: {e}")
        return

    # Overlapping window info
    start = max(forecast_series.index.min(), event_data_df.index.min())
    end   = min(forecast_series.index.max(), event_data_df.index.max())

    print("=" * 62)
    print("Peak Timing Error Inspector")
    print(f"  Lead day         : {lead}")
    print(f"  Forecast series  : {series_label}  [{forecast_key}]")
    print("=" * 62)

    print(f"\nOverlapping window:")
    print(f"  Start : {start}")
    print(f"  End   : {end}")
    n_steps = len(forecast_series.loc[start:end].dropna())
    print(f"  Steps : {n_steps}")

    print(f"\nObserved peak:")
    print(f"  Time     : {t_obs}")
    print(f"  Flow     : {q_obs:.3f} m³/s")

    print(f"\nForecast peak  [{series_label}]:")
    print(f"  Time     : {t_fcst}")
    print(f"  Flow     : {q_fcst:.3f} m³/s")

    print(f"\nCalculation:")
    print(f"  Δt = t_peak_forecast − t_peak_observed")
    print(f"     = {t_fcst}")
    print(f"       − {t_obs}")
    diff_sign = "early" if delta_t < 0 else ("late" if delta_t > 0 else "exact")
    print(f"\n  ┌──────────────────────────────────────────────────┐")
    print(f"  │  Δt_peak = {delta_t:+.1f} hours  ({diff_sign})              │")
    print(f"  └──────────────────────────────────────────────────┘")

    print(f"\nNote on magnitude:")
    ratio = q_fcst / q_obs if q_obs > 0 else np.nan
    print(f"  Forecast peak / Observed peak = {q_fcst:.1f} / {q_obs:.1f} = {ratio:.3f}")
    print(f"  (Timing error is independent of this ratio)")
````

````python
def compute_threshold_crossing(forecast_series, observed_df, obs_thresholds, sim_thresholds, rp_levels=None):
    """
    For each return period threshold, finds the first ascending crossing
    in both the observed and forecast series within the overlapping window,
    then computes Δt = t_crossing_forecast − t_crossing_observed (hours).

    'First ascending crossing' = the first timestep where the series
    transitions from below the threshold to at-or-above it.

    Parameters
    ----------
    forecast_series : pd.Series
    observed_df     : pd.DataFrame  (1 column)
    obs_thresholds  : dict  {rp: threshold_value}  — for observed series
    sim_thresholds  : dict  {rp: threshold_value}  — for forecast series
    rp_levels       : list of int, default [2,5,10,25,50,100]

    Returns
    -------
    results : dict  {rp: {'crossed_obs': bool, 'crossed_fcst': bool,
                           't_obs': Timestamp or None,
                           't_fcst': Timestamp or None,
                           'delta_t': float or None}}
    """
    if rp_levels is None:
        rp_levels = [2, 5, 10, 25, 50, 100]

    # Overlapping window
    start = max(forecast_series.index.min(), observed_df.index.min())
    end   = min(forecast_series.index.max(), observed_df.index.max())

    if start >= end:
        return {rp: {'crossed_obs': False, 'crossed_fcst': False,
                     't_obs': None, 't_fcst': None, 'delta_t': None}
                for rp in rp_levels}

    obs_win  = observed_df.iloc[:, 0].loc[start:end].dropna()
    fcst_win = forecast_series.loc[start:end].dropna()

    def first_ascending_crossing(series, threshold):
        """Returns timestamp of first ascending crossing, or None."""
        above = series >= threshold
        # Ascending: False→True transition
        crossings = above & (~above.shift(1, fill_value=False))
        idx = crossings[crossings].index
        return idx[0] if len(idx) > 0 else None

    results = {}
    for rp in rp_levels:
        t_obs  = first_ascending_crossing(obs_win,  obs_thresholds[rp])
        t_fcst = first_ascending_crossing(fcst_win, sim_thresholds[rp])

        crossed_obs  = t_obs  is not None
        crossed_fcst = t_fcst is not None

        if crossed_obs and crossed_fcst:
            delta_t = (t_fcst - t_obs).total_seconds() / 3600.0
        else:
            delta_t = None

        results[rp] = {
            'crossed_obs' : crossed_obs,
            'crossed_fcst': crossed_fcst,
            't_obs'       : t_obs,
            't_fcst'      : t_fcst,
            'delta_t'     : delta_t,
        }

    return results
````

````python
def compute_crossing_all_members(lead_forecasts_dict, event_data_df, observed_rperiods_df, simulated_rperiods_df, comid, rp_levels=None, max_lead=15):
    """
    Computes first ascending threshold crossing timing error for every
    ensemble member, statistic, lead day, and return period threshold.

    Returns
    -------
    crossing_results : dict
        {lead: {series_key: {rp: {'delta_t': float or None,
                                   'crossed_obs': bool,
                                   'crossed_fcst': bool}}}}
    all_keys : list
    """
    if rp_levels is None:
        rp_levels = [2, 5, 10, 25, 50, 100]

    obs_thresholds = {rp: observed_rperiods_df.loc[rp, comid] for rp in rp_levels}
    sim_thresholds = {rp: simulated_rperiods_df.loc[rp, comid] for rp in rp_levels}

    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    crossing_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing threshold crossings"):

        if lead not in lead_forecasts_dict:
            crossing_results[lead] = {
                k: {rp: {'delta_t': None, 'crossed_obs': False,
                          'crossed_fcst': False} for rp in rp_levels}
                for k in all_keys
            }
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        crossing_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                crossing_results[lead][key] = compute_threshold_crossing(
                    forecast_series, event_data_df,
                    obs_thresholds, sim_thresholds, rp_levels)
            except (ValueError, KeyError):
                crossing_results[lead][key] = {
                    rp: {'delta_t': None, 'crossed_obs': False,
                         'crossed_fcst': False}
                    for rp in rp_levels
                }

    return crossing_results, all_keys
````

````python
def get_crossing_vals(lead, rp, include_hires):
    """
    Returns Δt values and names ONLY for members that crossed both
    observed and forecast thresholds. Members that did not cross are
    counted separately.
    """
    keys = member_keys_c + ([hires_key_c] if include_hires else [])
    crossed_vals, crossed_names = [], []
    n_missed  = 0  # obs crossed, forecast did not
    n_no_obs  = 0  # obs never crossed (threshold too high for this event)

    for k in keys:
        r = crossing_results.get(lead, {}).get(k, {}).get(rp, {})
        if r.get('crossed_obs') and r.get('crossed_fcst'):
            dt = r.get('delta_t')
            if dt is not None:
                crossed_vals.append(dt)
                crossed_names.append(k)
        elif r.get('crossed_obs') and not r.get('crossed_fcst'):
            n_missed += 1
        elif not r.get('crossed_obs'):
            n_no_obs += 1

    return (np.array(crossed_vals), crossed_names, n_missed, n_no_obs, len(keys))
````

````python
def crps_ensemble_timestep(ensemble_array, obs_array):
    """
    Computes CRPS and its two components at every timestep using the
    energy score decomposition (Gneiting & Raftery, 2007):

        CRPS(t) = E|X - obs|  −  ½·E|X - X'|

    where X and X' are independent draws from the ensemble.

    The two terms have direct physical interpretation:
        MAE component  = E|X - obs|     — penalizes distance from obs
        Spread         = ½·E|X - X'|   — rewards ensemble spread
                                          (subtracted from CRPS)

    A perfectly calibrated ensemble with good resolution minimizes CRPS
    by having large spread exactly when needed (near the observation).

    Parameters
    ----------
    ensemble_array : np.ndarray  (n_timesteps, n_members)
        Ensemble forecast values. May contain NaN for missing members.
    obs_array : np.ndarray  (n_timesteps,)
        Observed values. NaN timesteps are skipped.

    Returns
    -------
    crps     : np.ndarray (n_timesteps,)  — CRPS per timestep
    mae_comp : np.ndarray (n_timesteps,)  — MAE component (reliability proxy)
    spread   : np.ndarray (n_timesteps,)  — spread component (resolution proxy)

    References
    ----------
    Gneiting, T. & Raftery, A.E. (2007). Strictly proper scoring rules,
        prediction, and estimation. JASA, 102(477), 359–378.
    Hersbach, H. (2000). Decomposition of the continuous ranked probability
        score for ensemble prediction systems. WAF, 15(5), 559–570.
    """
    T, M      = ensemble_array.shape
    crps      = np.full(T, np.nan)
    mae_comp  = np.full(T, np.nan)
    spread    = np.full(T, np.nan)

    for t in range(T):
        obs   = obs_array[t]
        x     = ensemble_array[t]
        valid = x[~np.isnan(x)]

        if len(valid) == 0 or np.isnan(obs):
            continue

        mae_comp[t] = np.mean(np.abs(valid - obs))
        # Mean pairwise absolute difference (vectorized)
        spread[t]   = np.mean(np.abs(valid[:, None] - valid[None, :])) / 2
        crps[t]     = mae_comp[t] - spread[t]

    return crps, mae_comp, spread


def compute_crps_by_lead(lead_forecasts_dict, event_data_df, max_lead=15):
    """
    Computes mean CRPS, MAE component, and spread component for each
    lead day over the overlapping window between forecast and observations.

    Unlike MCC/HSS/KGE, CRPS evaluates the FULL ENSEMBLE at each timestep —
    there is no per-member value. The result is a single scalar per lead day.

    Parameters
    ----------
    lead_forecasts_dict : dict
    event_data_df       : pd.DataFrame  (observed streamflow, 1 column)
    max_lead            : int

    Returns
    -------
    crps_results : dict
        {lead: {'crps': float, 'mae_component': float, 'spread': float,
                'n_timesteps': int}}
    """
    obs_series = event_data_df.iloc[:, 0]
    crps_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing CRPS per lead"):

        if lead not in lead_forecasts_dict:
            crps_results[lead] = {
                'crps': np.nan, 'mae_component': np.nan,
                'spread': np.nan, 'n_timesteps': 0}
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)

        # Use only ensemble members 01–51 (not statistics, not ensemble_52)
        member_cols = [c for c in lead_df.columns
                       if c.startswith("ensemble_") and c != "ensemble_52"]

        # Align temporal window
        start = max(lead_df.index.min(), obs_series.index.min())
        end   = min(lead_df.index.max(), obs_series.index.max())

        if start >= end:
            crps_results[lead] = {
                'crps': np.nan, 'mae_component': np.nan,
                'spread': np.nan, 'n_timesteps': 0}
            continue

        fcst_window = lead_df.loc[start:end, member_cols]
        obs_window  = obs_series.loc[start:end]

        # Align on common index
        common_idx  = fcst_window.index.intersection(obs_window.index)
        if len(common_idx) == 0:
            crps_results[lead] = {
                'crps': np.nan, 'mae_component': np.nan,
                'spread': np.nan, 'n_timesteps': 0}
            continue

        ens_arr = fcst_window.loc[common_idx].to_numpy(dtype=float)
        obs_arr = obs_window.loc[common_idx].to_numpy(dtype=float)

        crps_t, mae_t, spread_t = crps_ensemble_timestep(ens_arr, obs_arr)

        # Average over valid timesteps
        valid = ~np.isnan(crps_t)
        crps_results[lead] = {
            'crps'         : float(np.mean(crps_t[valid]))   if valid.any() else np.nan,
            'mae_component': float(np.mean(mae_t[valid]))    if valid.any() else np.nan,
            'spread'       : float(np.mean(spread_t[valid])) if valid.any() else np.nan,
            'n_timesteps'  : int(valid.sum()),
        }

    return crps_results
````

````python
def compute_correlation(forecast_series, observed_df):
    """
    Computes Pearson correlation coefficient (r) between a forecast
    series and observations within the overlapping temporal window.

    Parameters
    ----------
    forecast_series : pd.Series
    observed_df     : pd.DataFrame  (1 column)

    Returns
    -------
    r : float  in [-1, 1], or np.nan if insufficient data
    """
    start = max(forecast_series.index.min(), observed_df.index.min())
    end   = min(forecast_series.index.max(), observed_df.index.max())

    if start >= end:
        return np.nan

    fcst = forecast_series.loc[start:end].dropna()
    obs  = observed_df.iloc[:, 0].loc[start:end].dropna()

    common = fcst.index.intersection(obs.index)
    if len(common) < 3:
        return np.nan

    fcst_c = fcst.loc[common].values
    obs_c  = obs.loc[common].values

    if np.std(fcst_c) == 0 or np.std(obs_c) == 0:
        return np.nan

    r = np.corrcoef(fcst_c, obs_c)[0, 1]
    return float(r)


def compute_correlation_all_members(lead_forecasts_dict, event_data_df, max_lead=15):
    """
    Computes Pearson r for every ensemble member and statistic
    at every lead day.

    Returns
    -------
    r_results : dict  {lead: {series_key: r_value}}
    all_keys  : list
    """
    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    r_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing r per lead"):

        if lead not in lead_forecasts_dict:
            r_results[lead] = {k: np.nan for k in all_keys}
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        r_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                r_results[lead][key] = compute_correlation(
                    forecast_series, event_data_df)
            except (ValueError, KeyError):
                r_results[lead][key] = np.nan

    return r_results, all_keys
````

````python
# ===========================================================================
# Definition:
#   β = μ_forecast / μ_observed
#
#   β = 1   → no bias
#   β < 1   → forecast underestimates (systematic low bias)
#   β > 1   → forecast overestimates (systematic high bias)
# ===========================================================================

def compute_bias_ratio(forecast_series, observed_df):
    """
    Computes the bias ratio (β) between a forecast series and
    observations within the overlapping temporal window.

        β = μ_forecast / μ_observed

    Parameters
    ----------
    forecast_series : pd.Series
    observed_df     : pd.DataFrame  (1 column)

    Returns
    -------
    beta : float  or np.nan if insufficient data or μ_observed = 0
    """
    start = max(forecast_series.index.min(), observed_df.index.min())
    end   = min(forecast_series.index.max(), observed_df.index.max())

    if start >= end:
        return np.nan

    fcst = forecast_series.loc[start:end].dropna()
    obs  = observed_df.iloc[:, 0].loc[start:end].dropna()

    common = fcst.index.intersection(obs.index)
    if len(common) < 2:
        return np.nan

    mu_fcst = fcst.loc[common].mean()
    mu_obs  = obs.loc[common].mean()

    if mu_obs == 0:
        return np.nan

    return float(mu_fcst / mu_obs)


def compute_bias_ratio_all_members(lead_forecasts_dict, event_data_df, max_lead=15):
    """
    Computes bias ratio (β) for every ensemble member and statistic
    at every lead day.

    Returns
    -------
    beta_results : dict  {lead: {series_key: beta_value}}
    all_keys     : list
    """
    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    beta_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing β per lead"):

        if lead not in lead_forecasts_dict:
            beta_results[lead] = {k: np.nan for k in all_keys}
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        beta_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                beta_results[lead][key] = compute_bias_ratio(
                    forecast_series, event_data_df)
            except (ValueError, KeyError):
                beta_results[lead][key] = np.nan

    return beta_results, all_keys
````

````python
# ===========================================================================
# Definition (Kling et al., 2012 — KGE'):
#   γ = CV_forecast / CV_observed
#     = (σ_forecast / μ_forecast) / (σ_observed / μ_observed)
#
#   γ = 1   → variability ratio perfect (relative spread matches)
#   γ < 1   → forecast underestimates variability
#   γ > 1   → forecast overestimates variability
#
# Advantage over α = σ_f/σ_o: γ is independent of bias (β).
# A model can have β << 1 (severe underestimation) but γ = 1
# (correct relative variability), which α would not reveal.
#
# Reference:
#   Kling, H., Fuchs, M., & Paulin, M. (2012). Runoff conditions in the
#   upper Danube basin under an ensemble of climate change scenarios.
#   Journal of Hydrology, 424–425, 264–277.
# ===========================================================================

def compute_variability_ratio(forecast_series, observed_df):
    """
    Computes the variability ratio (γ) between a forecast series and
    observations within the overlapping temporal window.

        γ = CV_forecast / CV_observed
          = (σ_forecast / μ_forecast) / (σ_observed / μ_observed)

    Parameters
    ----------
    forecast_series : pd.Series
    observed_df     : pd.DataFrame  (1 column)

    Returns
    -------
    gamma : float  or np.nan if insufficient data, zero mean, or zero std
    """
    start = max(forecast_series.index.min(), observed_df.index.min())
    end   = min(forecast_series.index.max(), observed_df.index.max())

    if start >= end:
        return np.nan

    fcst = forecast_series.loc[start:end].dropna()
    obs  = observed_df.iloc[:, 0].loc[start:end].dropna()

    common = fcst.index.intersection(obs.index)
    if len(common) < 3:
        return np.nan

    fcst_c = fcst.loc[common].values
    obs_c  = obs.loc[common].values

    mu_fcst = np.mean(fcst_c)
    mu_obs  = np.mean(obs_c)
    sd_fcst = np.std(fcst_c, ddof=1)
    sd_obs  = np.std(obs_c,  ddof=1)

    # Cannot compute CV if mean or std is zero
    if mu_fcst == 0 or mu_obs == 0 or sd_obs == 0:
        return np.nan

    cv_fcst = sd_fcst / mu_fcst
    cv_obs  = sd_obs  / mu_obs

    if cv_obs == 0:
        return np.nan

    return float(cv_fcst / cv_obs)


def compute_variability_ratio_all_members(lead_forecasts_dict, event_data_df, max_lead=15):
    """
    Computes variability ratio (γ) for every ensemble member and statistic
    at every lead day.

    Returns
    -------
    gamma_results : dict  {lead: {series_key: gamma_value}}
    all_keys      : list
    """
    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    gamma_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing γ per lead"):

        if lead not in lead_forecasts_dict:
            gamma_results[lead] = {k: np.nan for k in all_keys}
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        gamma_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                gamma_results[lead][key] = compute_variability_ratio(
                    forecast_series, event_data_df)
            except (ValueError, KeyError):
                gamma_results[lead][key] = np.nan

    return gamma_results, all_keys
````

````python
def display_variability_ratio_table(forecast_key):
    """
    Shows a table of variability ratio (γ) values across all lead days
    for the selected forecast series, including the CV components.
    """
    series_label = next(
        (lbl for lbl, key in selectable_options if key == forecast_key),
        forecast_key
    )

    obs_series = event_data_df.iloc[:, 0]

    print("=" * 65)
    print("Variability Ratio (γ) — KGE' component  (Kling et al., 2012)")
    print(f"  Forecast series : {series_label}  [{forecast_key}]")
    print(f"  γ = CV_forecast / CV_observed")
    print(f"  γ = 1 → perfect  |  γ < 1 → underestimates variability  |  γ > 1 → overestimates")
    print("=" * 65)

    rows = []
    for lead in range(16):
        gamma = gamma_results.get(lead, {}).get(forecast_key, np.nan)

        cv_f_str = "—"
        cv_o_str = "—"

        if lead in lead_forecasts_dict:
            try:
                lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
                fs      = resolve_forecast_series(lead_df, forecast_key)
                start   = max(fs.index.min(), obs_series.index.min())
                end     = min(fs.index.max(), obs_series.index.max())
                common  = fs.loc[start:end].dropna().index.intersection(
                          obs_series.loc[start:end].dropna().index)
                if len(common) >= 3:
                    fcst_c  = fs.loc[common].values
                    obs_c   = obs_series.loc[common].values
                    mu_f    = np.mean(fcst_c)
                    mu_o    = np.mean(obs_c)
                    sd_f    = np.std(fcst_c, ddof=1)
                    sd_o    = np.std(obs_c,  ddof=1)
                    if mu_f != 0 and mu_o != 0:
                        cv_f_str = f"{sd_f/mu_f:.4f}"
                        cv_o_str = f"{sd_o/mu_o:.4f}"
            except Exception:
                pass

        if np.isnan(gamma):
            gamma_str = "N/A"
            interp    = "—"
        else:
            gamma_str = f"{gamma:.4f}"
            if abs(gamma - 1.0) < 0.05:
                interp = "Perfect"
            elif gamma < 1.0:
                pct    = (1 - gamma) * 100
                interp = f"Underestimates variability ({pct:.1f}%)"
            else:
                pct    = (gamma - 1) * 100
                interp = f"Overestimates variability ({pct:.1f}%)"

        rows.append({
            "Lead day"       : lead,
            "CV forecast"    : cv_f_str,
            "CV observed"    : cv_o_str,
            "γ"              : gamma_str,
            "Interpretation" : interp,
        })

    table = pd.DataFrame(rows).set_index("Lead day")
    display(HTML(table.to_html()))

    # Summary
    all_g = [gamma_results[l].get(forecast_key, np.nan)
             for l in range(16)
             if not np.isnan(gamma_results[l].get(forecast_key, np.nan))]
    if all_g:
        print(f"\nAcross all leads:")
        print(f"  Max γ  : {max(all_g):.4f}")
        print(f"  Min γ  : {min(all_g):.4f}")
        print(f"  Mean γ : {np.mean(all_g):.4f}")
        avg_pct = (np.mean(all_g) - 1) * 100
        direction = "underestimates" if avg_pct < 0 else "overestimates"
        print(f"  On average the forecast {direction} variability by {abs(avg_pct):.1f}%")
        print(f"\nNote: γ is independent of bias (β). A model with β << 1")
        print(f"      can still have γ = 1 if its relative variability is correct.")

def get_member_gamma(lead, include_hires):
    keys  = member_keys_g + ([hires_key_g] if include_hires else [])
    vals  = [gamma_results[lead][k] for k in keys
             if not np.isnan(gamma_results[lead].get(k, np.nan))]
    names = [k for k in keys
             if not np.isnan(gamma_results[lead].get(k, np.nan))]
    return np.array(vals), names
````

````python
# ===========================================================================
# Definition (Kling et al., 2012 — KGE'):
#   KGE' = 1 − √((r−1)² + (β−1)² + (γ−1)²)
#
#   where:
#     r = Pearson correlation          (computed by compute_correlation)
#     β = μ_forecast / μ_observed      (computed by compute_bias_ratio)
#     γ = CV_forecast / CV_observed    (computed by compute_variability_ratio)
#
#   KGE' = 1    → perfect forecast
#   KGE' > −0.41 → better than mean flow benchmark
#   KGE' = −∞   → worst possible
#
# Reference:
#   Kling, H., Fuchs, M., & Paulin, M. (2012). Runoff conditions in the
#   upper Danube basin under an ensemble of climate change scenarios.
#   Journal of Hydrology, 424–425, 264–277.
# ===========================================================================

def compute_kge(forecast_series, observed_df):
    """
    Computes KGE' (Kling et al., 2012) and its three components
    within the overlapping temporal window.

        KGE' = 1 − √((r−1)² + (β−1)² + (γ−1)²)

    Parameters
    ----------
    forecast_series : pd.Series
    observed_df     : pd.DataFrame  (1 column)

    Returns
    -------
    kge   : float  or np.nan
    r     : float  Pearson correlation
    beta  : float  bias ratio
    gamma : float  variability ratio (CV-based)
    """
    r     = compute_correlation(forecast_series, observed_df)
    beta  = compute_bias_ratio(forecast_series, observed_df)
    gamma = compute_variability_ratio(forecast_series, observed_df)

    if any(np.isnan(v) for v in [r, beta, gamma]):
        return np.nan, r, beta, gamma

    kge = 1.0 - np.sqrt((r - 1)**2 + (beta - 1)**2 + (gamma - 1)**2)
    return float(kge), float(r), float(beta), float(gamma)


def compute_kge_all_members(lead_forecasts_dict, event_data_df, max_lead=15):
    """
    Computes KGE' and its three components for every ensemble member
    and statistic at every lead day.

    Returns
    -------
    kge_results : dict
        {lead: {series_key: {'kge': float, 'r': float,
                              'beta': float, 'gamma': float}}}
    all_keys    : list
    """
    lead0_df = build_lead_dataframe(lead_forecasts_dict, 0)
    all_keys = [key for _, key in get_forecast_options(lead0_df)
                if key is not None]

    kge_results = {}

    for lead in tqdm(range(0, max_lead + 1), desc="Computing KGE' per lead"):

        if lead not in lead_forecasts_dict:
            kge_results[lead] = {
                k: {'kge': np.nan, 'r': np.nan, 'beta': np.nan, 'gamma': np.nan}
                for k in all_keys
            }
            continue

        lead_df = build_lead_dataframe(lead_forecasts_dict, lead)
        kge_results[lead] = {}

        for key in all_keys:
            try:
                forecast_series = resolve_forecast_series(lead_df, key)
                kge, r, beta, gamma = compute_kge(forecast_series, event_data_df)
                kge_results[lead][key] = {
                    'kge'  : kge,
                    'r'    : r,
                    'beta' : beta,
                    'gamma': gamma,
                }
            except (ValueError, KeyError):
                kge_results[lead][key] = {
                    'kge': np.nan, 'r': np.nan,
                    'beta': np.nan, 'gamma': np.nan,
                }

    return kge_results, all_keys
````

````python
def display_kge_table(forecast_key):
    """
    Shows a table of KGE' and its components across all lead days
    for the selected forecast series.
    """
    series_label = next(
        (lbl for lbl, key in selectable_options if key == forecast_key),
        forecast_key
    )

    print("=" * 62)
    print("KGE'  (Kling et al., 2012)  — All lead days")
    print(f"  Forecast series : {series_label}  [{forecast_key}]")
    print("=" * 62)

    rows = []
    for lead in range(16):
        d   = kge_results.get(lead, {}).get(forecast_key, {})
        kge = d.get('kge',   np.nan)
        r   = d.get('r',     np.nan)
        b   = d.get('beta',  np.nan)
        g   = d.get('gamma', np.nan)

        def fmt(v): return f"{v:.4f}" if not np.isnan(v) else "N/A"

        if np.isnan(kge):
            interp = "—"
        elif kge > 0.75:
            interp = "Good"
        elif kge > 0.50:
            interp = "Intermediate"
        elif kge > 0.00:
            interp = "Poor"
        elif kge > -0.41:
            interp = "Very poor"
        else:
            interp = "Unacceptable"

        rows.append({
            "Lead": lead,
            "r"   : fmt(r),
            "β"   : fmt(b),
            "γ"   : fmt(g),
            "KGE'": fmt(kge),
            "Interpretation": interp,
        })

    table = pd.DataFrame(rows).set_index("Lead")
    display(HTML(table.to_html()))

    all_kge = [kge_results[l].get(forecast_key, {}).get('kge', np.nan)
               for l in range(16)
               if not np.isnan(kge_results[l].get(forecast_key, {}).get('kge', np.nan))]
    if all_kge:
        best_lead = [l for l in range(16)
                     if abs(kge_results[l].get(forecast_key, {}).get('kge', -999)
                            - max(all_kge)) < 1e-6][0]
        print(f"\nAcross all leads:")
        print(f"  Best KGE'  : {max(all_kge):.4f}  (lead {best_lead})")
        print(f"  Worst KGE' : {min(all_kge):.4f}")
        print(f"  Mean KGE'  : {np.mean(all_kge):.4f}")
        print(f"\n  Benchmark (mean flow): KGE' = −0.41")
        n_above = sum(1 for v in all_kge if v > -0.41)
        print(f"  Leads above benchmark: {n_above} / {len(all_kge)}")

def get_member_kge(lead, include_hires):
    keys  = member_keys_kge + ([hires_key_kge] if include_hires else [])
    vals  = [kge_results[lead][k]['kge'] for k in keys
             if not np.isnan(kge_results[lead].get(k, {}).get('kge', np.nan))]
    names = [k for k in keys
             if not np.isnan(kge_results[lead].get(k, {}).get('kge', np.nan))]
    return np.array(vals), names
````

# 2. The Dual-Threshold Design

## 2.1 Motivation

A fundamental challenge in verifying global hydrological forecasts is that the model's **climatology** differs from the observed climatology. GEOGLOWS (like GloFAS) produces streamflow using a hydrological model forced by numerical weather prediction output. The return period thresholds derived from this simulated climatology are systematically different from those derived from observed gauge records.

If the observed 2-year return period is 82 m³/s but the simulated 2-year return period is 33 m³/s, classifying both the observation and the forecast against the same threshold would introduce a systematic bias into every categorical metric.

## 2.2 Implementation

This framework uses a **dual-threshold approach**:

$$\text{Observed categories} \leftarrow \text{Thresholds from historical gauge data (HIS)}$$

$$\text{Forecast categories} \leftarrow \text{Thresholds from GEOGLOWS retrospective simulation}$$

This means:
- Observations are classified using return periods computed from the observed annual maxima series, fitted with a Gumbel distribution
- Forecasts are classified using return periods computed from the GEOGLOWS retrospective daily simulation (`retro_daily`), also fitted with a Gumbel distribution

## 2.3 Return period categories

The contingency matrix uses the following categories, defined dynamically based on the magnitude of the observed event:

| Category label | Observed threshold | Forecast threshold |
|---|---|---|
| < 2yr | Flow < $Q_{2,obs}$ | Flow < $Q_{2,sim}$ |
| 2–5yr | $Q_{2,obs} \leq$ Flow $< Q_{5,obs}$ | $Q_{2,sim} \leq$ Flow $< Q_{5,sim}$ |
| 5–10yr | $Q_{5,obs} \leq$ Flow $< Q_{10,obs}$ | ... |
| ... | ... | ... |
| ≥100yr | Flow $\geq Q_{100,obs}$ | Flow $\geq Q_{100,sim}$ |

The number of categories is determined by the maximum return period exceeded during the observed event.

# 3. Forecast Reorganization by Lead Time

## 3.1 The challenge: multiple initialization dates

GEOGLOWS produces a new ensemble forecast every day. For a flood event spanning, say, 4 days (28–31 October), the system issues forecasts initialized on each of the preceding 15 days plus the event days themselves. Each initialization produces a 15-day ensemble forecast covering a different temporal window.

To verify forecast skill **as a function of lead time**, all these forecasts must be reorganized. Instead of asking "what did the forecast initialized on October 26 predict?", the verification framework asks "what did all forecasts at 3 days of lead time predict for this event?"

## 3.2 Lead time definition

The lead time $d$ of a forecast timestep $t$ with respect to initialization time $t_0$ is defined using ceiling-based 24-hour windows:

$$d = \left\lceil \frac{t - t_0}{24\text{ h}} \right\rceil$$

This means:
- **Lead 0**: the initialization timestep only ($t = t_0$)
- **Lead 1**: the interval $(0\text{ h},\ 24\text{ h}]$ after initialization — the first forecast day
- **Lead 2**: the interval $(24\text{ h},\ 48\text{ h}]$ — the second forecast day
- ...
- **Lead $d$**: the interval $((d-1) \times 24\text{ h},\ d \times 24\text{ h}]$

## 3.3 The reorganization procedure

The function `reorganize_forecasts_by_daily_lead` iterates over all initialization dates and assigns each forecast timestep to its corresponding lead day bucket. The result is a nested dictionary:

```
lead_forecasts_dict[lead_day][init_date] = {
    "time_3h" : timestamps of 3-hourly ensemble members,
    "ens_3h"  : array of shape (n_timesteps, 51),
    "time_1h" : timestamps of high-resolution member,
    "ens_1h"  : array of shape (m_timesteps, 1)
}
```

For a given lead day, `build_lead_dataframe` then concatenates all initialization dates into a single DataFrame. This produces the data structure that feeds all verification metrics.

## 3.4 The two ensemble resolutions

GEOGLOWS provides two types of forecast output:

| Member | Temporal resolution | Forecast horizon | Count |
|---|---|---|---|
| Ensemble members 01–51 | 3-hourly | 15 days | 51 |
| High-resolution (ensemble_52) | 1-hourly | 10 days | 1 |

Before reorganization, the 3-hourly ensemble members are **interpolated to hourly resolution** using linear interpolation. This ensures temporal alignment with the high-resolution member and with the observed gauge data (which is hourly in this case).

> **Note on interpolation**: linear interpolation fills the intervals between 3-hourly values by assuming a straight line between consecutive points. This cannot produce values higher or lower than the original 3-hourly values — the peak of the interpolated series is identical to the peak of the native 3-hourly output. The interpolation serves purely to align the ensemble temporally with the hourly observed record and with the high-resolution member (ensemble_52).

## 3.5 Initialization window

For each event, the forecast download covers a window of **15 days prior to the event start** through the event end date. This ensures that all lead times from 0 to 15 days have at least one initialization date contributing data.

The number of initialization dates depends on the duration of the event. For an event spanning $D$ days, the download window covers $15 + D$ days, yielding $15 + D$ initialization dates. Longer events produce more initialization dates and therefore larger sample sizes within each lead day bucket.

## 3.6 Important interpretation caveat: initialization pooling

When `build_lead_dataframe` concatenates all initialization dates for a given lead day, a given observed timestep may appear **more than once** in the resulting DataFrame — once for each initialization date whose lead-$d$ window covers that timestep.

This happens because the event window is covered by multiple overlapping forecasts. The flood peak, for instance, falls within the lead-3 window of one initialization, the lead-4 window of the next, the lead-5 window of the one before that, and so on. All of these forecast values are compared against the same observed value at that timestep.

# 4. Categorical Metrics

## 4.1 The Contingency Matrix

The multi-category contingency matrix $\mathbf{C}$ is a $K \times K$ table where $K$ is the number of return period categories. Element $C_{ij}$ counts the number of timesteps where:
- The **observation** fell in category $i$ (row)
- The **forecast** fell in category $j$ (column)

$$\mathbf{C} = \begin{pmatrix} C_{11} & C_{12} & \cdots & C_{1K} \\ C_{21} & C_{22} & \cdots & C_{2K} \\ \vdots & & \ddots & \vdots \\ C_{K1} & C_{K2} & \cdots & C_{KK} \end{pmatrix}$$

A perfect forecast system produces non-zero values **only on the diagonal** ($C_{ii}$). Off-diagonal elements indicate forecast errors:

- **Lower triangle** ($C_{ij}$ where $i > j$): the observation was in a higher category than the forecast → **underestimation**
- **Upper triangle** ($C_{ij}$ where $i < j$): the observation was in a lower category than the forecast → **overestimation**

### 4.1.1 Reading the matrix

The matrix in this Colab shows **raw timestep counts** (not fractions). For a given lead day and forecast series:

$$\text{Hits} = \sum_{i} C_{ii} \quad \text{(diagonal sum)}$$

$$\text{Underestimation} = \sum_{i > j} C_{ij} \quad \text{(lower triangle)}$$

$$\text{Overestimation} = \sum_{i < j} C_{ij} \quad \text{(upper triangle)}$$

$$N = \sum_{i,j} C_{ij} \quad \text{(total timesteps)}$$

### 4.1.2 Important note on "misses"

Following the WMO/WWRP verification framework and Hewson (2007), this Colab treats all off-diagonal elements (both underestimation and overestimation) as errors, analogous to "misses + false alarms" in a binary contingency table.

## 4.2 Matthews Correlation Coefficient (MCC)

The multi-category MCC (Gorodkin, 2004; Jurman et al., 2012) generalizes the binary MCC to $K$ categories:

$$\text{MCC} = \frac{N \cdot c - \sum_k t_k p_k}{\sqrt{\left(N^2 - \sum_k p_k^2\right)\left(N^2 - \sum_k t_k^2\right)}}$$

Every term in this formula comes directly from the contingency matrix $\mathbf{C}$:

- $N$ — total number of timesteps (sum of all cells in the matrix)
- $c = \sum_k C_{kk}$ — total hits: the sum of the **diagonal** cells only
- $t_k$ — row sum of row $k$: how many timesteps were **observed** in category $k$
- $p_k$ — column sum of column $k$: how many timesteps were **forecast** in category $k$
- $\sum_k t_k p_k$ — sum of products of each row total by its corresponding column total

The term $\sum_k t_k p_k$ deserves a closer look. It represents the number of hits that a **random forecast** would produce on average — not a uniformly random one, but one that assigns categories with the same overall frequencies as the actual forecast. For example, if 80% of observed timesteps are `<2yr` and 75% of forecast timesteps are also `<2yr`, then by pure chance alone one would expect $0.80 \times 0.75 = 60\%$ of cases to match in that category. $\sum_k t_k p_k$ sums this expected overlap across all categories.

The numerator is therefore the **excess hits above chance**: how many more correct classifications the forecast achieves compared to what random assignment would produce. The denominator normalizes this excess so the result always falls between $-1$ and $+1$.

### 4.2.1 Interpretation

| Range | Interpretation |
|---|---|
| MCC = 1 | Perfect agreement between forecast and observation categories |
| MCC > 0 | Forecast is better than random assignment |
| MCC = 0 | No better than random |
| MCC < 0 | Systematic disagreement (forecast and observation categories are anticorrelated) |

### 4.2.2 A key property for flood events

MCC is sensitive to how the observations are distributed across categories. During a flood event, most timesteps fall in the `<2yr` category (normal flow), and only a few fall in the extreme categories. A model that **always predicts `<2yr`** will still have many hits — but its MCC will be close to 0, because the denominator accounts for the marginal totals and penalizes this behavior.

This is desirable: a forecast that only predicts "nothing unusual" should not receive credit for skill it does not have.

## 4.3 Heidke Skill Score (HSS)

The multi-category HSS answers the question: **by how much does the forecast improve over what a random forecast would achieve?**

$$\text{HSS} = \frac{N \cdot c - \sum_k t_k p_k}{N^2 - \sum_k t_k p_k}$$

The terms are the same as in MCC:

- $N \cdot c - \sum_k t_k p_k$ — the numerator: actual hits above what chance would produce
- $N^2 - \sum_k t_k p_k$ — the denominator: the maximum possible improvement above chance

A useful equivalent form makes the interpretation explicit:

$$\text{HSS} = \frac{PC - PC_{\text{ref}}}{1 - PC_{\text{ref}}}$$

where:
- $PC = c / N$ is the **proportion correct** — the fraction of timesteps correctly classified
- $PC_{\text{ref}} = \sum_k t_k p_k / N^2$ is the **proportion correct expected by chance** — what a random forecast with the same category frequencies would achieve

HSS is therefore the fraction of the gap between chance and perfection that the forecast actually closes.

### 4.3.1 Interpretation

| HSS value | Meaning |
|---|---|
| 1 | Perfect forecast |
| > 0 | Better than a random forecast |
| 0 | No better than chance |
| < 0 | Worse than chance |

### 4.3.2 MCC vs HSS: which one to trust?

Both metrics share the **same numerator**. The only difference is in the denominator, and this matters during flood events.

MCC normalizes by the geometric mean of the spread in the row totals and the spread in the column totals. HSS normalizes by a simpler quantity that is dominated by the most frequent category.

In practice this means: when most timesteps fall in `<2yr` (normal flow), **HSS gives more credit to a model that correctly classifies those normal timesteps**, while **MCC is more demanding and requires the model to also classify the rare extreme categories correctly**.

Neither metric is universally "better." The recommendation is to report both:
- If MCC and HSS agree → the conclusion is robust regardless of which metric you use
- If MCC is much lower than HSS → the model is achieving apparent skill mainly by correctly predicting normal flow, not the extreme event. In this case, MCC is the more informative metric for flood verification purposes.

# 5. Timing Metrics

## 5.1 Peak Timing Error ($\Delta t_\text{peak}$)

$$\Delta t_\text{peak} = t_\text{peak,forecast} - t_\text{peak,observed} \quad [\text{hours}]$$

where $t_\text{peak}$ is the timestamp of maximum streamflow within the overlapping window between forecast and observations.

- $\Delta t_\text{peak} < 0$: forecast peak arrives **early**
- $\Delta t_\text{peak} > 0$: forecast peak arrives **late**
- $\Delta t_\text{peak} = 0$: perfect timing

**$\Delta t_\text{peak}$ is independent of magnitude.** A member that predicts a peak of 80 m³/s at exactly the right time has $\Delta t = 0$ even if the observed peak was 9,609 m³/s. This independence is intentional and operationally valuable: timing accuracy and magnitude accuracy are distinct aspects of forecast quality that should be assessed separately.

**Interpretation changes with lead time.** At short leads (1–3 days), the atmospheric forcing is well-constrained and timing errors should be small. At long leads (10–15 days), larger timing errors are expected as uncertainty propagates through the meteorological–hydrological chain.

**Recommendations**
- Use the **median** $\Delta t$ across members as the primary summary statistic
- Use the **IQR** as a measure of ensemble timing spread
- Negative median $\Delta t$ (early bias) is often preferable operationally to late bias, since early warnings give communities more preparation time

## 5.2 Threshold Crossing Timing Error

For a given return period threshold $Q_{RP}$, the threshold crossing timing error is:

$$\Delta t_{RP} = t_\text{crossing,forecast} - t_\text{crossing,observed} \quad [\text{hours}]$$

where $t_\text{crossing}$ is the timestamp of the **first ascending crossing** of the threshold — the first timestep at which streamflow transitions from below to at-or-above the threshold.

Flood early warning systems issue alerts based on **threshold exceedance**, not peak arrival. A warning is triggered when the forecast probability of exceeding $Q_{2yr}$ or $Q_{5yr}$ crosses a decision threshold. Therefore, knowing when the model first detects the crossing of each return period level is more directly relevant to operational decision-making than peak timing.

Many ensemble members may **never cross** a given threshold, especially for extreme events with severe model underestimation. This Colab reports two quantities separately:

1. **Detection rate**: fraction of members that crossed both the observed and forecast thresholds (and thus have a computable $\Delta t_{RP}$)
2. **Conditional timing error**: $\Delta t_{RP}$ computed only for members that did detect the crossing

These are fundamentally different pieces of information. A high detection rate with large $\Delta t$ means the model consistently detects the event but late. A low detection rate means the model mostly fails to detect the event at all — in which case timing error is secondary.

As the threshold increases (from 2yr to 100yr), the detection rate typically decreases because fewer members generate flows large enough to cross the higher thresholds. This pattern reveals the **limits of the ensemble's probabilistic coverage** of extreme events.

# 6. Accuracy Metrics

## 6.1 Kling-Gupta Efficiency (KGE')

The Nash-Sutcliffe Efficiency (NSE) and its variants have long been the standard for hydrological model assessment, but they are sensitive to extreme values and confound multiple sources of error. Gupta et al. (2009) decomposed the MSE into three orthogonal components, and Kling et al. (2012) refined this decomposition into what is now called KGE'.

$$\text{KGE}' = 1 - \sqrt{(r-1)^2 + (\beta-1)^2 + (\gamma-1)^2}$$

where the three components are:

**Correlation** $r$: Pearson correlation coefficient between forecast and observation

$$r = \frac{\sum_t (Q_{f,t} - \bar{Q}_f)(Q_{o,t} - \bar{Q}_o)}{\sqrt{\sum_t (Q_{f,t} - \bar{Q}_f)^2 \cdot \sum_t (Q_{o,t} - \bar{Q}_o)^2}}$$

**Bias ratio** $\beta$: ratio of forecast mean to observed mean

$$\beta = \frac{\bar{Q}_f}{\bar{Q}_o}$$

**Variability ratio** $\gamma$: ratio of forecast coefficient of variation to observed coefficient of variation (Kling et al., 2012)

$$\gamma = \frac{CV_f}{CV_o} = \frac{\sigma_f / \bar{Q}_f}{\sigma_o / \bar{Q}_o}$$

### 6.1.1 Interpretation of each component

| Component | Perfect value | > Perfect | < Perfect |
|---|---|---|---|
| $r$ | 1 | N/A (bounded at 1) | Poor timing / shape |
| $\beta$ | 1 | Overestimation bias | Underestimation bias |
| $\gamma$ | 1 | Overestimates variability | Underestimates variability |

The **dominant error source** is the component furthest from 1, i.e., the largest of $(r-1)^2$, $(\beta-1)^2$, $(\gamma-1)^2$.

### 6.1.2 Performance classification

| KGE' range | Category |
|---|---|
| 0.75 < KGE' ≤ 1.00 | Good |
| 0.50 < KGE' ≤ 0.75 | Intermediate |
| 0.00 < KGE' ≤ 0.50 | Poor |
| −0.41 < KGE' ≤ 0.00 | Very poor |
| KGE' ≤ −0.41 | Unacceptable |

The benchmark value of $\text{KGE}' = -0.41$ corresponds to using the observed mean flow as the forecast for every timestep. A model below this benchmark provides no useful information beyond climatology.

## 6.2 Continuous Ranked Probability Score (CRPS)

The CRPS evaluates the ensemble as a **probability distribution**, not member by member. It measures the integrated squared difference between the empirical CDF of the ensemble and the step function at the observed value:

$$\text{CRPS} = \int_{-\infty}^{\infty} \left[F(x) - \mathbf{1}(x \geq Q_{obs})\right]^2 dx$$

where $F(x)$ is the empirical CDF of the ensemble members and $\mathbf{1}(x \geq Q_{obs})$ is the Heaviside step function at the observation.


### 6.2.1 Energy score decomposition

For a discrete ensemble of $M$ members, the CRPS is computed via the energy score formula (Gneiting & Raftery, 2007):

$$\text{CRPS} = \underbrace{\frac{1}{M}\sum_{m=1}^{M}|X_m - Q_{obs}|}_{\text{MAE component}} - \underbrace{\frac{1}{2M^2}\sum_{m=1}^{M}\sum_{m'=1}^{M}|X_m - X_{m'}|}_{\text{Spread}}$$

The two terms have direct physical interpretations:

**MAE component** $= \mathbb{E}|X - Q_{obs}|$: the mean absolute error of the ensemble members against the observation. This penalizes systematic bias.

**Spread** $= \frac{1}{2}\mathbb{E}|X - X'|$: half the mean pairwise absolute difference between ensemble members. This rewards ensemble spread — and is subtracted from the CRPS because a well-dispersed ensemble should be penalized less than a deterministic forecast.

Therefore:

$$\text{CRPS} = \text{MAE component} - \text{Spread}$$

The MAE component is always $\geq$ the CRPS. The spread component represents the "discount" earned by the ensemble for having internal variability.

### 6.2.2 Key properties

- **Units**: same as the variable being forecast (m³/s). This makes CRPS directly interpretable in physical terms
- **Perfect score**: CRPS = 0 (achieved by a perfect deterministic forecast or a perfectly calibrated ensemble with zero spread)
- **Lower is better**: unlike KGE' or MCC, CRPS has no upper bound for poor forecasts
- **Proper scoring rule**: a forecaster cannot game the CRPS by over- or under-dispersing the ensemble. The optimal strategy is always to issue honest probability estimates

### 6.2.3 CRPS as a function of lead time

Unlike KGE' and MCC (which are computed per member), CRPS produces a **single value per lead day**, representing the ensemble's collective performance averaged over all timesteps in the overlapping window.

An important artifact: because CRPS is averaged over all timesteps (including the many normal-flow periods), extreme peak timesteps are diluted. The CRPS may appear "moderate" even for a catastrophic failure at the flood peak, because normal-flow timesteps contribute small CRPS values that bring the mean down.
